```python
import matplotlib.pyplot as plt
# import matplotlib.ticker as ticker
import pandas as pd

fig = plt.figure()
ax1 = fig.add_subplot(121)
ax2 = fig.add_subplot(122)

data_tail = pd.read_csv(f'perf_tail_100.txt', sep='\\s+', comment='#')
data_notail = pd.read_csv(f'perf_notail_100.txt', sep='\\s+', comment='#')
ax1.plot(data_tail['InBER'], data_tail['BER'], label='Tailing zeros, n=100')
ax1.plot(data_notail['InBER'], data_notail['BER'], label='No tail, n=100')
ax2.plot(data_tail['SNR'], data_tail['BER'], label='Tailing zeros, n=100')
ax2.plot(data_notail['SNR'], data_notail['BER'], label='No tail, n=100')

data_tail = pd.read_csv(f'perf_tail_500.txt', sep='\\s+', comment='#')
data_notail = pd.read_csv(f'perf_notail_500.txt', sep='\\s+', comment='#')
ax1.plot(data_tail['InBER'], data_tail['BER'], label='Tailing zeros, n=500')
ax1.plot(data_notail['InBER'], data_notail['BER'], label='No tail, n=500')
ax2.plot(data_tail['SNR'], data_tail['BER'], label='Tailing zeros, n=500')
ax2.plot(data_notail['SNR'], data_notail['BER'], label='No tail, n=500')

ax1.set(
	title = 'Viterbi decoding, R=1/2',
    xscale = 'log',
	yscale = 'log',
	xlabel = 'InBER',
	ylabel = 'OutBER',
)
ax2.set(
	title = 'Viterbi decoding, R=1/2',
    yscale = 'log',
	xlabel = 'SNR as BPSK',
	ylabel = 'OutBER',
)
gridparams = {
    'which': 'major',
	'color': [0.7, 0.7, 0.7],
	'linestyle': '-',
}
ax1.grid(gridparams)
ax2.grid(gridparams)
ax1.invert_xaxis()
ax1.legend()
ax2.legend()
fig.set_figwidth(20)
fig.set_figheight(10)
plt.show()

```
